### 🤔 What is Agentic RAG?

**Traditional RAG:**
```
User Question → Retrieve Documents → Generate Answer
```

**Agentic RAG:**
```
User Question → AI Agent analyzes question → Retrieves relevant documents → Agent processes and synthesizes → Intelligent Answer
```

**Key Difference:** Instead of just retrieving and generating, **intelligent agents** understand context, make decisions, and provide more thoughtful responses.

````python
# 🔧 Setup and Configuration
import os
from typing import List, Dict, Optional
from dataclasses import dataclass
from dotenv import load_dotenv

# Import Semantic Kernel components
from semantic_kernel import Kernel
from semantic_kernel.connectors.ai.open_ai import AzureChatCompletion, AzureTextEmbedding
from semantic_kernel.connectors.ai.function_choice_behavior import FunctionChoiceBehavior

# Load environment variables
load_dotenv()

# Create kernel for our Agentic RAG system
kernel = Kernel()

# Configure Azure OpenAI services
chat_completion = AzureChatCompletion(
    service_id="agentic_rag_chat",
    api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    deployment_name=os.getenv("AZURE_OPENAI_DEPLOYMENT_NAME"),
)
kernel.add_service(chat_completion)

text_embedding = AzureTextEmbedding(
    service_id="agentic_rag_embedding",
    api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    deployment_name=os.getenv("AZURE_OPENAI_ADA_DEPLOYMENT")
)
kernel.add_service(text_embedding)

# Configure execution settings
execution_settings = kernel.get_prompt_execution_settings_from_service_id("agentic_rag_chat")
execution_settings.function_choice_behavior = FunctionChoiceBehavior.Auto()
execution_settings.max_tokens = 1500
execution_settings.temperature = 0.3

print("✅ Agentic RAG System initialized")
print("🔗 Azure OpenAI services connected")
print("⚙️ Execution settings configured")
````

## 🤖 Step 3: Create Smart RAG Agent

Now let's create a simple but intelligent RAG agent that can analyze questions, retrieve relevant documents, and provide thoughtful responses.

````python
class SmartRAGAgent:
    """A simple but intelligent RAG agent that can analyze questions and provide thoughtful answers."""
    
    def __init__(self, kernel: Kernel, document_store: List[SimpleDocument]):
        self.kernel = kernel
        self.document_store = document_store
        self.conversation_history = []
    
    async def answer_question(self, question: str, max_documents: int = 3) -> Dict:
        """Answer a question using intelligent document retrieval and analysis."""
        
        print(f"🤔 **Question:** {question}")
        print("─" * 50)
        
        # Step 1: Analyze the question
        question_analysis = await self._analyze_question(question)
        print(f"🧠 **Question Type:** {question_analysis['type']}")
        print(f"🎯 **Key Topics:** {', '.join(question_analysis['topics'])}")
        
        # Step 2: Retrieve relevant documents
        relevant_docs = await self._smart_retrieval(question, max_documents)
        print(f"📚 **Found {len(relevant_docs)} relevant documents**")
        
        # Step 3: Generate intelligent response
        response = await self._generate_response(question, relevant_docs, question_analysis)
        
        # Step 4: Store in conversation history
        self.conversation_history.append({
            "question": question,
            "analysis": question_analysis,
            "documents_used": [doc.title for doc in relevant_docs],
            "response": response
        })
        
        return {
            "question": question,
            "answer": response,
            "source_documents": [{"title": doc.title, "category": doc.category} for doc in relevant_docs],
            "confidence": "High" if len(relevant_docs) >= 2 else "Medium"
        }
    
    async def _analyze_question(self, question: str) -> Dict:
        """Analyze the question to understand intent and topics."""
        
        # Simple question analysis based on keywords
        question_lower = question.lower()
        
        # Determine question type
        if any(word in question_lower for word in ["what", "define", "explain"]):
            q_type = "explanation"
        elif any(word in question_lower for word in ["how", "steps", "process"]):
            q_type = "how-to"
        elif any(word in question_lower for word in ["compare", "difference", "vs"]):
            q_type = "comparison"
        else:
            q_type = "factual"
        
        # Extract key topics (simple keyword extraction)
        semantic_keywords = ["semantic", "kernel", "plugin", "function", "agent", "embedding", "memory"]
        topics = [keyword for keyword in semantic_keywords if keyword in question_lower]
        
        if not topics:
            topics = ["general"]
        
        return {
            "type": q_type,
            "topics": topics,
            "focus": question
        }
    
    async def _smart_retrieval(self, question: str, max_documents: int) -> List[SimpleDocument]:
        """Retrieve the most relevant documents using embedding similarity."""
        
        # Generate embedding for the question
        question_embedding = await text_embedding.generate_embeddings([question])
        question_vector = question_embedding[0]
        
        # Calculate similarity with all documents
        similarities = []
        for doc in self.document_store:
            if doc.embedding is not None:
                # Simple cosine similarity
                similarity = self._calculate_similarity(question_vector, doc.embedding)
                similarities.append((doc, similarity))
        
        # Sort by similarity and return top documents
        similarities.sort(key=lambda x: x[1], reverse=True)
        return [doc for doc, score in similarities[:max_documents]]
    
    def _calculate_similarity(self, vec1, vec2) -> float:
        """Calculate cosine similarity between two vectors."""
        import numpy as np
        
        # Convert to numpy arrays if needed
        vec1 = np.array(vec1)
        vec2 = np.array(vec2)
        
        # Calculate cosine similarity
        dot_product = np.dot(vec1, vec2)
        magnitude1 = np.linalg.norm(vec1)
        magnitude2 = np.linalg.norm(vec2)
        
        if magnitude1 * magnitude2 == 0:
            return 0
        return dot_product / (magnitude1 * magnitude2)
    
    async def _generate_response(self, question: str, docs: List[SimpleDocument], analysis: Dict) -> str:
        """Generate an intelligent response based on retrieved documents."""
        
        # Combine relevant document content
        context = "\n\n".join([f"**{doc.title}**: {doc.content}" for doc in docs])
        
        response_prompt = f"""
        Based on the following information about Semantic Kernel, provide a helpful and accurate answer to the user's question.
        
        Context Information:
        {context}
        
        Question: {question}
        Question Type: {analysis['type']}
        Key Topics: {', '.join(analysis['topics'])}
        
        Instructions:
        1. Answer the question directly and clearly
        2. Use information from the provided context
        3. Be concise but comprehensive
        4. If the context doesn't contain enough information, acknowledge this
        5. For how-to questions, provide step-by-step guidance when possible
        
        Answer:
        """
        
        try:
            # Simple response generation using kernel
            response = await self.kernel.invoke_prompt(
                prompt=response_prompt
            )
            
            return str(response)
            
        except Exception as e:
            # Fallback response if AI generation fails
            return f"Based on the available information about {', '.join(analysis['topics'])}, here's what I found: {'. '.join([doc.content[:100] + '...' for doc in docs])}"

# Create our smart RAG agent
rag_agent = SmartRAGAgent(kernel, document_store)

print("✅ Smart RAG Agent created!")
print("✅ Ready to answer questions intelligently")
print("✅ Agent can analyze questions, retrieve documents, and synthesize responses")
print("\n🎯 **What makes this 'Agentic'?**")
print("   • 🧠 Analyzes questions to understand intent")
print("   • 🔍 Performs intelligent document retrieval")  
print("   • 💡 Synthesizes information from multiple sources")
print("   • 📝 Provides structured, thoughtful responses")
print("   • 🧠 Maintains conversation context")
print("\n🚀 Let's test our agent!")
````
