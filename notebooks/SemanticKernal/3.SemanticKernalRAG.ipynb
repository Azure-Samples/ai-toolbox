### Example: Create a Simple Native Plugin

Let's create a plugin with a native Python function:

````python
from semantic_kernel.functions import kernel_function

# Create a simple plugin class
class TextPlugin:
    @kernel_function(
        name="get_word_count",
        description="Counts the number of words in a text"
    )
    def get_word_count(self, text: str) -> str:
        word_count = len(text.split())
        return f"The text contains {word_count} words."
    
    @kernel_function(
        name="uppercase",
        description="Converts text to uppercase"
    )
    def to_uppercase(self, text: str) -> str:
        return text.upper()

# Add the plugin to the kernel
kernel.add_plugin(TextPlugin(), plugin_name="TextPlugin")

print("✓ TextPlugin added to kernel")
````

## 📚 Step 4: Document Loading & Text Splitting Strategy

Now we move from prompt engineering to **document processing** - the foundation of any RAG system!

### 🧠 Understanding Document Processing in RAG

**The Critical Role of Document Processing:**

RAG systems are only as good as their document processing strategy. Poor chunking leads to:
- ❌ **Irrelevant retrieval** - Wrong chunks returned for queries
- ❌ **Lost context** - Important information split across chunks  
- ❌ **Poor attribution** - Difficulty tracing answers to sources
- ❌ **Degraded performance** - LLM struggles with fragmented information

### 🏗️ The RAG Indexing Pipeline

Before we can retrieve, we need to **index our documents**:

```
📄 Raw Documents → 🔪 Text Splitting → 🧮 Embedding Generation → 🗄️ Vector Storage
```

This happens **offline** (once) to prepare your knowledge base for real-time querying.

### 🎯 Text Splitting Strategies

**Chunking Strategy Comparison:**

| Strategy | Pros | Cons | Best For |
|----------|------|------|----------|
| **Fixed Length** | Simple, predictable | Breaks context mid-sentence | Uniform content |
| **Sentence-based** | Natural boundaries | Variable chunk sizes | General text |
| **Paragraph-based** | Preserves logical structure | Can be too large/small | Well-formatted documents |
| **Semantic Chunking** | Content-aware splits | Complex implementation | Technical documentation |
| **Recursive Character** | Balances context & size | More complex logic | Mixed content types |

### 📏 Chunk Size Considerations

**The Chunk Size Dilemma:**

**Smaller Chunks (100-200 tokens):**
- ✅ More precise retrieval
- ✅ Better semantic matching  
- ❌ May lose context
- ❌ Requires more chunks for complex answers

**Larger Chunks (500-1000 tokens):**
- ✅ Preserves context and relationships
- ✅ Fewer chunks needed per query
- ❌ Less precise retrieval
- ❌ May include irrelevant information

**Our Strategy:** We'll use **paragraph-based chunking** which naturally preserves logical content boundaries while maintaining reasonable chunk sizes.

1. **Load** document content (using mock AI-focused documents)
2. **Split** it into logical paragraphs using natural boundaries
3. **Validate** chunk quality and size distribution
4. **Prepare** for embedding generation in the next step

### 💡 Production Considerations

**For Real Applications, Consider:**
- **Document types** - PDFs, HTML, Word docs require different processing
- **Metadata preservation** - Keep source info, page numbers, sections
- **Content cleaning** - Remove headers, footers, navigation elements
- **Language detection** - Handle multilingual content appropriately
- **Update strategies** - Incremental updates vs full reprocessing

## 🧮 Step 5: Vector Embeddings & Semantic Kernel's InMemoryStore

Now comes the magic! Let's convert our text chunks into **vector embeddings** and store them in Semantic Kernel's **InMemoryStore** for lightning-fast semantic search.

### 🔬 Understanding Vector Embeddings

**What Are Vector Embeddings?**

Vector embeddings are numerical representations of text that capture semantic meaning:

```
"Machine learning uses algorithms" → [0.234, -0.567, 0.890, ..., 0.123]
                                      ↑ 1,536 floating-point numbers
```

**Key Properties:**
- ✅ **Semantic similarity** - Similar meanings = similar vectors
- ✅ **Mathematical operations** - Enable similarity calculations
- ✅ **High dimensionality** - Capture complex semantic relationships
- ✅ **Language understanding** - Work across languages and domains

### 🎯 How Semantic Search Works

**Traditional Keyword Search:**
```
Query: "AI learning algorithms"
Match: Documents containing exact words "AI", "learning", "algorithms"
Problem: Misses "machine learning", "neural networks", "deep learning"
```

**Semantic Vector Search:**
```
Query: "AI learning algorithms" → Vector: [0.123, -0.456, ...]
↓ Calculate similarity scores
Document: "Machine learning uses neural networks" → Score: 0.87
Document: "Pizza recipes and cooking tips" → Score: 0.23
Result: Returns most semantically relevant content!
```

### Step 5.2: Define Document Model with Vector Store Decorator

SK uses dataclasses with special decorators to define document models:

````python
from dataclasses import dataclass
from typing import Annotated
from semantic_kernel.data.vector import VectorStoreField, vectorstoremodel

@vectorstoremodel(collection_name="documents")
@dataclass
class DocumentParagraph:
    """Data model for storing document paragraphs with embeddings"""
    id: Annotated[str, VectorStoreField("key")]
    text: Annotated[str, VectorStoreField("data")]
    embedding: Annotated[
        list[float] | None,
        VectorStoreField(
            "vector",
            dimensions=1536,
            # We'll set the embedding generator later to avoid scope issues
        ),
    ] = None

print("✓ DocumentParagraph model defined")
print("  - Fields: id (key), text (data), embedding (vector)")
print("  - Auto-embedding will be enabled via collection configuration")
print("  - Dimensions: 1536 (Azure OpenAI text-embedding-ada-002)")
````

### 🔍 Understanding Vector Embeddings (Interactive Demo)

**🎯 What This Cell Does:**
Before we move to RAG queries, let's see vector embeddings in action! This interactive demo shows how:

1. **Similar Meanings = Similar Vectors**: Sentences with similar meanings have high similarity scores
2. **Different Topics = Different Vectors**: Unrelated sentences have lower similarity scores  
3. **Semantic Understanding**: The AI understands meaning, not just keywords

**🧮 How It Works:**
- We convert 3 test sentences into vectors using the same embedding model
- We calculate **cosine similarity** between the vectors (0-1 scale, higher = more similar)
- You'll see that similar meanings score ~0.85-0.95, different topics score ~0.60-0.75

**💡 Why This Matters:**
This is exactly how RAG search works - your query gets converted to a vector, then we find the most similar document vectors!

Run this cell to see vector similarity in action:

````python
import numpy as np

def cosine_similarity(vec_a: list[float], vec_b: list[float]) -> float:
    """Calculate cosine similarity between two vectors (0-1, higher = more similar)"""
    return np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))

# Test sentences with different similarity levels
test_sentences = [
    "Machine learning uses algorithms to learn from data",
    "Algorithms enable computers to learn from experience",  # Similar meaning
    "Pizza and pasta are Italian foods"  # Completely different topic
]

# Generate embeddings for test sentences
print("Generating embeddings for test sentences...\n")
test_embeddings = []
for sentence in test_sentences:
    response = await text_embedding.generate_embeddings([sentence])
    test_embeddings.append(response[0])

# Compare first sentence with the other two
base_sentence = test_sentences[0]
print(f"Base sentence: '{base_sentence}'\n")
print("Similarity comparisons:")
print("-" * 80)

for i in range(1, len(test_sentences)):
    similarity = cosine_similarity(test_embeddings[0], test_embeddings[i])
    print(f"📊 Comparison sentence: '{test_sentences[i]}'")
    print(f"   Similarity score: {similarity:.4f}")
    print()

print("💡 Notice: Similar meanings have higher scores (closer to 1.0)")
print("   Even when different words are used!")
````

**What Just Happened?**

1. We converted three sentences into vectors (1,536 numbers each)
2. We used **cosine similarity** to measure how "close" the meanings are
3. The second sentence scores high (~0.85-0.95) because it means nearly the same thing
4. The third sentence scores low (~0.60-0.75) because it's about a completely different topic

**This is how vector search works:**
- Your query gets converted to a vector
- The system finds the most similar document vectors
- Similar meaning = high similarity score = relevant results!

### Step 6.2: Create Search Function (Memory Recall)

**🎯 What This Cell Does:**
This cell creates a custom search function that the LLM can call to find relevant information. Here's what happens:

1. **Custom Plugin**: We create a `MemoryPlugin` class with a `recall` function
2. **Query Processing**: When called, it converts the query into an embedding vector
3. **Similarity Search**: Compares the query vector with all document vectors using cosine similarity
4. **Return Results**: Returns the top 3 most relevant document chunks as text

**🔧 Technical Implementation:**
- Uses manual vector similarity calculation (more reliable than auto-search)
- Employs numpy for efficient cosine similarity computation
- Returns formatted text that the LLM can use as context

**🚀 Why This Approach:**
We use a custom implementation to avoid embedding generation issues and ensure reliable search results.

Run this cell to create your search function:

````python
# Create a custom search function using manual similarity calculation
import numpy as np
from semantic_kernel.functions import kernel_function

class MemoryPlugin:
    @kernel_function(
        name="recall",
        description="Recalls information from the document collection about AI topics."
    )
    async def recall(self, query: str) -> str:
        # Generate embedding for the query
        query_embedding = await text_embedding.generate_embeddings([query])
        query_vector = query_embedding[0]
        
        # Get all documents from collection and calculate similarities manually
        all_docs = []
        similarities = []
        
        for doc_item in document_items:
            doc_vector = np.array(doc_item.embedding)
            # Calculate cosine similarity
            similarity = np.dot(query_vector, doc_vector) / (np.linalg.norm(query_vector) * np.linalg.norm(doc_vector))
            similarities.append((similarity, doc_item.text))
        
        # Sort by similarity and get top 3
        similarities.sort(key=lambda x: x[0], reverse=True)
        top_results = similarities[:3]
        
        # Extract text from top results
        context_texts = [result[1] for result in top_results]
        
        return "\n\n".join(context_texts)

# Add the memory plugin to the kernel
memory_plugin = MemoryPlugin()
kernel.add_plugin(memory_plugin, plugin_name="memory")

print("✓ Search function 'memory.recall' added to kernel")
print("  - Function performs manual vector similarity search")
print("  - Returns top 3 most relevant document chunks")
print("  - Uses cosine similarity for ranking")
````
